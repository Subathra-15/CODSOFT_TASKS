Importing the libraries

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

Loading the dataset

```python
df=pd.read_csv("train_data.txt",sep=" ::: ",engine="python",names=["ID","TITLE",'GENRE',"DESCRIPTION"])
```

```python
df
```

```python
X=df['DESCRIPTION']
y=df['GENRE']
```

```python
print(y.unique())
```

Splitting intro training and testing set

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

Vectorization

```python
tfidf=TfidfVectorizer(stop_words='english',max_features=5000)
X_train_tfidf=tfidf.fit_transform(X_train)
X_test_tfidf=tfidf.transform(X_test)
```

Train the model and make predictions

```python
svc=LinearSVC()
svc.fit(X_train_tfidf,y_train)
y_pred=svc.predict(X_test_tfidf)
```

```python
print(y_pred)
```

```python
print(y_test)
```

Test the model using a different dataset

Load the dataset

```python
test_df=pd.read_csv("test_data.txt",sep=" ::: ",engine="python",names=["ID","TITLE","DESCRIPTION"])
```

```python
print(test_df)
```

```python
X_testing=test_df['DESCRIPTION'].fillna('')
```

```python
X_testing_tfidf=tfidf.transform(X_testing)
```

```python
test_predictions=svc.predict(X_testing_tfidf)
```

Printing the predictions of the test data

```python
print(
    np.concatenate((np.array(X_testing).reshape(len(test_predictions),1),
                   np.array(test_predictions).reshape(len(test_predictions),1))
                   ,1)
)
```
