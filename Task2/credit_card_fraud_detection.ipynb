Importing the libraries

```python
import numpy as np
import pandas as pd
```

Loading the dataset

```python
df=pd.read_csv("fraudTrain.csv")
```

```python
df=df.dropna()
```

```python
df.head()
```

Seperate X and y

```python
X=df.drop(['is_fraud','Unnamed: 0'],axis=1)
y=df['is_fraud']
```

Convert date columns

```python
X['trans_date_trans_time']=pd.to_datetime(X['trans_date_trans_time'])
X['hour']=X['trans_date_trans_time'].dt.hour
X['day']=X['trans_date_trans_time'].dt.day
X['month']=X['trans_date_trans_time'].dt.month
X['dob']=pd.to_datetime(X['dob'])
X['age']=2026-X['dob'].dt.year
X.drop(['trans_date_trans_time','dob'],axis=1,inplace=True)
```

Label Encode only text columns

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
for col in X.select_dtypes(include="object").columns:
  X[col]=le.fit_transform(X[col].astype(str))
```

Scale everything

```python
from sklearn.preprocessing import StandardScaler
s=StandardScaler()
X=s.fit_transform(X)
```

Training the model

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(X,y)
```
